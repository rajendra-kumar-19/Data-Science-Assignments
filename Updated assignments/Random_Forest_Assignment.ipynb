# Additional Notes

## 1. Explain Bagging and Boosting methods. How is it different from each other.
Bagging:
Bagging (Bootstrap Aggregating) trains multiple models independently on different random samples of the dataset and combines their predictions. It helps reduce variance and overfitting.

Boosting:
Boosting trains models sequentially, where each new model focuses on correcting the mistakes of the previous model. It helps improve accuracy and reduce bias.

Difference Between Bagging and Boosting:
Bagging trains multiple models independently, while Boosting trains models sequentially.
Bagging reduces variance, while Boosting reduces bias.
Bagging uses random bootstrap samples, while Boosting focuses more on previously misclassified samples.
Bagging is faster because models can be trained in parallel, while Boosting is slower because models depend on previous models.

## 2. Explain how to handle imbalance in the data
Methods to handle imbalance:

Oversampling: Increase the number of minority class samples (e.g., SMOTE).
Undersampling: Reduce the number of majority class samples.
Class Weights: Assign higher weights to minority class observations during model training.
Synthetic Data Generation: Create new minority class samples using techniques like SMOTE.
Use Appropriate Evaluation Metrics: Use Precision, Recall, F1-Score, and ROC-AUC instead of only Accuracy.
