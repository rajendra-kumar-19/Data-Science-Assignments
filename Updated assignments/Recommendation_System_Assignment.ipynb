# Recommendation System Assignment

```python

# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python

# Load Dataset

df = pd.read_csv('anime.csv')

print(df.head())

print(df.shape)
```

```python

# Dataset Information

print(df.info())

print(df.isnull().sum())
```

```python

# Remove Missing Values

df.dropna(inplace=True)

print(df.shape)
```

```python

# Top Anime Genres

print(df['genre'].value_counts().head(10))
```

```python

# Visualization

df['genre'].value_counts().head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top Anime Genres')

plt.show()
```

```python

# Feature Selection

features = df['genre']

print(features.head())
```

```python

# Convert Text Data into Numerical Form

cv = CountVectorizer(stop_words='english')

feature_matrix = cv.fit_transform(features)

print(feature_matrix.shape)
```

```python

# Cosine Similarity Matrix

similarity = cosine_similarity(feature_matrix)

print(similarity.shape)
```

```python

# Recommendation Function

indices = pd.Series(df.index, index=df['name']).drop_duplicates()

def recommend_anime(anime_name):

    idx = indices[anime_name]

    sim_scores = list(enumerate(similarity[idx]))

    sim_scores = sorted(
        sim_scores,
        key=lambda x: x[1],
        reverse=True
    )

    sim_scores = sim_scores[1:6]

    anime_indices = [i[0] for i in sim_scores]

    return df['name'].iloc[anime_indices]

print(recommend_anime('Naruto'))
```

```python

# Threshold Experimentation

threshold_1 = similarity[0][1]

threshold_2 = similarity[0][2]

print('Threshold Example 1:', threshold_1)

print('Threshold Example 2:', threshold_2)
```


# Evaluation and Analysis

- Cosine similarity helps identify similar anime.
- Recommendation quality improves with better feature selection.
- Threshold experimentation changes recommendation strictness.
- Content-based filtering works well for anime recommendations.
