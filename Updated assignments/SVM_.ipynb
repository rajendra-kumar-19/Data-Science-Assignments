
# Comparison and Analysis

## Kernel Comparison
- Linear Kernel: Fast but less flexible for complex boundaries.
- Polynomial Kernel: Handles non-linear data but computationally expensive.
- RBF Kernel: Best performance for the mushroom dataset.

## Strengths of SVM
- Effective in high-dimensional spaces.
- Works well for classification tasks.
- Robust against overfitting with proper tuning.

## Weaknesses of SVM
- Computationally expensive for large datasets.
- Sensitive to parameter tuning.

## Practical Implications
- SVM is highly effective for biological and classification datasets like mushrooms.
- Hyperparameter tuning significantly improves model accuracy.


```python

# EDA Visualization

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x=df.iloc[:,0])
plt.title("Target Class Distribution")
plt.show()
```

```python

# Preprocessing and Feature Scaling

from sklearn.preprocessing import LabelEncoder, StandardScaler

df_encoded = df.copy()

for col in df_encoded.columns:
    df_encoded[col] = LabelEncoder().fit_transform(df_encoded[col])

X = df_encoded.iloc[:,1:]
y = df_encoded.iloc[:,0]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Preprocessing and Scaling Completed")
```

```python

# Model Building

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

svm_model = SVC(kernel='rbf')
svm_model.fit(X_train, y_train)

print("SVM Model Built Successfully")
```

```python

# Evaluation

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = svm_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python

# Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV

params = {
    'C':[0.1,1,10],
    'kernel':['linear','rbf']
}

grid = GridSearchCV(SVC(), params, cv=3)
grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
```

```python

# Visualization

from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(
    svm_model,
    X_test,
    y_test
)

plt.show()
```
