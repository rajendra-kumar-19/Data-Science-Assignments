# PCA 

```python

# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python

# Load Dataset

df = pd.read_csv('wine.csv')

print(df.head())
print(df.shape)
```

```python

# Check Missing Values

print(df.isnull().sum())
```

```python

# Summary Statistics

print(df.describe())
```

```python

# Histograms

df.hist(figsize=(12,10))
plt.show()
```

```python

# Correlation Heatmap

plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python

# Feature Scaling

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)
```

```python

# Apply PCA

pca = PCA()

pca_data = pca.fit_transform(scaled_data)

print(pca.explained_variance_ratio_)
```

```python

# Cumulative Explained Variance

plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.show()
```

```python

# PCA with 2 Components

pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(
    data=pca_result,
    columns=['PC1', 'PC2']
)

print(pca_df.head())
```

```python

# KMeans Clustering on Original Data

kmeans_original = KMeans(n_clusters=3, random_state=42)

original_clusters = kmeans_original.fit_predict(scaled_data)

print(original_clusters)
```

```python

# Silhouette Score for Original Data

original_score = silhouette_score(
    scaled_data,
    original_clusters
)

print('Original Data Silhouette Score:', original_score)
```

```python

# KMeans Clustering on PCA Data

kmeans_pca = KMeans(n_clusters=3, random_state=42)

pca_clusters = kmeans_pca.fit_predict(pca_result)

print(pca_clusters)
```

```python

# Silhouette Score for PCA Data

pca_score = silhouette_score(
    pca_result,
    pca_clusters
)

print('PCA Data Silhouette Score:', pca_score)
```

```python

# Cluster Visualization

plt.figure(figsize=(8,6))

plt.scatter(
    pca_result[:,0],
    pca_result[:,1],
    c=pca_clusters,
    cmap='viridis'
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Clustering')

plt.show()
```

## Visualization of Clustering on Original Data

```python

# Original Data Cluster Visualization (using first two scaled features)

plt.figure(figsize=(8,6))

plt.scatter(
    scaled_data[:,0],
    scaled_data[:,1],
    c=original_clusters,
    cmap='viridis'
)

plt.title("K-Means Clusters on Original Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```


## Discussion

- PCA was applied after feature scaling.
- K-Means clustering was performed on both original and PCA-transformed data.
- Silhouette scores were used to evaluate clustering quality.
- Visualization was generated for both original and PCA-transformed datasets.



## Comparison

| Method | Dataset | Evaluation |
|----------|----------|----------|
| K-Means | Original Scaled Data | Original Silhouette Score |
| K-Means | PCA Data | PCA Silhouette Score |

The clustering performance can be compared using silhouette scores obtained from both datasets.



## Conclusion

- PCA successfully reduced dimensionality while preserving most of the variance.
- K-Means clustering was applied before and after PCA.
- Silhouette scores were compared to evaluate clustering quality.
- Cluster visualizations were generated for both original and PCA-transformed data.
- PCA can improve visualization and may improve clustering performance depending on the dataset.
