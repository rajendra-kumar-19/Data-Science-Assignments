
# Estimation and Confidence Intervals

## Background
In quality control, destructive sampling is used to test high-value items. Since testing destroys the product, only small samples can be used.

## Scenario
A manufacturer of print-heads wants to estimate the mean durability of print-heads in millions of characters printed before failure.

## Sample Data
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29


```python

# Import Required Libraries
import numpy as np
import scipy.stats as stats
```

```python

# Sample Data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32,
        0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

# Sample Size
n = len(data)

# Sample Mean
sample_mean = np.mean(data)

# Sample Standard Deviation
sample_std = np.std(data, ddof=1)

print("Sample Size:", n)
print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
```


# Task A: 99% Confidence Interval Using Sample Standard Deviation

Since the population standard deviation is unknown and the sample size is small (n < 30), we use the **t-distribution**.

Formula:

CI = x̄ ± t * (s / √n)

Where:
- x̄ = sample mean
- s = sample standard deviation
- n = sample size
- t = t-critical value


```python

# Confidence Level
confidence = 0.99

# Degrees of Freedom
df = n - 1

# t-Critical Value
t_critical = stats.t.ppf((1 + confidence) / 2, df)

# Margin of Error
margin_error_t = t_critical * (sample_std / np.sqrt(n))

# Confidence Interval
lower_t = sample_mean - margin_error_t
upper_t = sample_mean + margin_error_t

print("t-Critical Value:", t_critical)
print("Margin of Error:", margin_error_t)

print("\n99% Confidence Interval using Sample Standard Deviation:")
print((lower_t, upper_t))
```


# Interpretation

We are 99% confident that the true mean durability of the print-heads lies within the calculated confidence interval.



# Task B: 99% Confidence Interval Using Known Population Standard Deviation

If the population standard deviation is known (σ = 0.2), we use the **z-distribution**.

Formula:

CI = x̄ ± z * (σ / √n)

Where:
- x̄ = sample mean
- σ = population standard deviation
- n = sample size
- z = z-critical value


```python

# Known Population Standard Deviation
population_std = 0.2

# z-Critical Value
z_critical = stats.norm.ppf((1 + confidence) / 2)

# Margin of Error
margin_error_z = z_critical * (population_std / np.sqrt(n))

# Confidence Interval
lower_z = sample_mean - margin_error_z
upper_z = sample_mean + margin_error_z

print("z-Critical Value:", z_critical)
print("Margin of Error:", margin_error_z)

print("\n99% Confidence Interval using Known Population Standard Deviation:")
print((lower_z, upper_z))
```


# Conclusion

1. When the population standard deviation is unknown and the sample size is small, the t-distribution is used.
2. When the population standard deviation is known, the z-distribution is used.
3. Both methods provide a range in which the true population mean is likely to lie with 99% confidence.
