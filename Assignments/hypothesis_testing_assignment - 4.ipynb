
# Hypothesis Testing Assignment


```

# Import Required Libraries
import numpy as np
from scipy.stats import norm
```


# Step 1: State the Hypotheses

- Null Hypothesis (H₀):
The weekly operating cost follows the existing cost model.

- Alternative Hypothesis (H₁):
The weekly operating costs are higher than the existing cost model.


```

# Given Data

sample_mean = 3050
X_mean = 600
sigma = 5 * 25
n = 25

# Theoretical Mean Cost
mu = 1000 + 5 * X_mean

print("Theoretical Mean Weekly Cost:", mu)
print("Sample Mean Weekly Cost:", sample_mean)
print("Population Standard Deviation:", sigma)
```

```

# Step 2: Calculate Test Statistic

z = (sample_mean - mu) / (sigma / np.sqrt(n))

print("Z-Test Statistic:", z)
```

```

# Step 3: Determine Critical Value

alpha = 0.05

critical_value = norm.ppf(1 - alpha)

print("Critical Value:", critical_value)
```

```

# Step 4: Make Decision

if z > critical_value:
    print("Reject the Null Hypothesis")
    print("There is strong evidence that weekly operating costs are higher.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence that weekly operating costs are higher.")
```


# Conclusion

The hypothesis testing process was completed using the Z-test. Based on the comparison of the test statistic and critical value, a conclusion was made regarding the restaurant owners' claim.
