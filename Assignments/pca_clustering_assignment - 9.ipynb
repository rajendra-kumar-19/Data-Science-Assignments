
# PCA and Clustering Analysis Assignment

This notebook performs Exploratory Data Analysis (EDA), Principal Component Analysis (PCA), and Clustering on the Wine dataset.


```

# Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```

# Load Dataset

df = pd.read_csv("wine.csv")

# Display first 5 rows
df.head()
```

```

# Dataset Information

print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Info:")
df.info()
```

```

# Check Missing Values

print(df.isnull().sum())
```


# Task 1: Exploratory Data Analysis (EDA)


```

# Statistical Summary

df.describe()
```

```

# Histograms

df.hist(figsize=(15,12), bins=15)

plt.tight_layout()
plt.show()
```

```

# Boxplots

plt.figure(figsize=(15,8))

sns.boxplot(data=df.select_dtypes(include=np.number))

plt.xticks(rotation=90)

plt.show()
```

```

# Correlation Heatmap

plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```


# Task 2: Dimensionality Reduction with PCA


```

# Standardization

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

scaled_df.head()
```

```

# Apply PCA

pca = PCA()

pca_data = pca.fit_transform(scaled_df)

# Explained Variance Ratio
explained_variance = pca.explained_variance_ratio_

print("Explained Variance Ratio:")
print(explained_variance)
```

```

# Scree Plot

plt.figure(figsize=(8,5))

plt.plot(range(1, len(explained_variance)+1),
         explained_variance.cumsum(),
         marker='o')

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")

plt.title("Scree Plot")

plt.show()
```

```

# Select 2 Principal Components

pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_df)

pca_df = pd.DataFrame(
    data=pca_result,
    columns=['PC1', 'PC2']
)

pca_df.head()
```


# Task 3: Clustering with Original Data


```

# K-Means Clustering on Original Data

kmeans_original = KMeans(n_clusters=3, random_state=42, n_init=10)

original_labels = kmeans_original.fit_predict(scaled_df)

# Evaluation Metrics
silhouette_original = silhouette_score(scaled_df, original_labels)

db_index_original = davies_bouldin_score(scaled_df, original_labels)

print("Silhouette Score (Original Data):", silhouette_original)

print("Davies-Bouldin Index (Original Data):", db_index_original)
```

```

# Visualize Clusters on Original Data

plt.figure(figsize=(8,6))

sns.scatterplot(
    x=scaled_df.iloc[:,0],
    y=scaled_df.iloc[:,1],
    hue=original_labels,
    palette='viridis'
)

plt.title("K-Means Clustering on Original Data")

plt.show()
```


# Task 4: Clustering with PCA Data


```

# K-Means Clustering on PCA Data

kmeans_pca = KMeans(n_clusters=3, random_state=42, n_init=10)

pca_labels = kmeans_pca.fit_predict(pca_df)

# Evaluation Metrics
silhouette_pca = silhouette_score(pca_df, pca_labels)

db_index_pca = davies_bouldin_score(pca_df, pca_labels)

print("Silhouette Score (PCA Data):", silhouette_pca)

print("Davies-Bouldin Index (PCA Data):", db_index_pca)
```

```

# Visualize Clusters on PCA Data

plt.figure(figsize=(8,6))

sns.scatterplot(
    x=pca_df['PC1'],
    y=pca_df['PC2'],
    hue=pca_labels,
    palette='Set1'
)

plt.title("K-Means Clustering on PCA Data")

plt.show()
```


# Task 5: Comparison and Analysis

## Observations

### Original Data
- Clustering was performed on the full-dimensional dataset.
- More information is retained but visualization is difficult.

### PCA Data
- PCA reduced the dimensionality to 2 components.
- Easier visualization and faster clustering.
- Some information loss may occur.

### Impact of PCA
- PCA simplifies the dataset while retaining maximum variance.
- It can improve clustering performance and visualization.
- Trade-off exists between dimensionality reduction and information retention.
