## Exercise 1: Prime Numbers

```python

# Program to check whether a number is prime or not

num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

## Exercise 2: Product of Random Numbers

```python

# Program to generate two random numbers and ask user for the product

import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print("Number 1:", num1)
print("Number 2:", num2)

user_answer = int(input("Enter the product of the two numbers: "))

correct_answer = num1 * num2

if user_answer == correct_answer:
    print("Correct! Well done.")
else:
    print("Incorrect.")
    print("Correct Answer is:", correct_answer)
```

## Exercise 3: Squares of Even Numbers

```python

# Program to print squares of even numbers from 100 to 200

# Choice: Even Numbers

for num in range(100, 201):
    if num % 2 == 0:
        print(f"Number: {num}, Square: {num**2}")
```

## Exercise 4: Word Counter

```python

# Program to count the number of words in a text

input_text = "This is a sample text. This text will be used to demonstrate the word counter."

words = input_text.split()

word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print("Word Count:")

for word, count in word_count.items():
    print(f"'{word}': {count}")
```

## Exercise 5: Check for Palindrome

```python

# Function to check whether a string is palindrome or not

import re

def is_palindrome(text):
    
    # Remove spaces, punctuation and convert to lowercase
    cleaned_text = re.sub(r'[^a-zA-Z0-9]', '', text).lower()
    
    # Check palindrome
    return cleaned_text == cleaned_text[::-1]

# Example
input_string = "racecar"

print("Input:", input_string)
print("Output:", is_palindrome(input_string))
```
