
# Comparison and Insights

## ARIMA Model
- Captures autocorrelation effectively.
- Suitable for stationary time series data.
- Requires parameter tuning.

## Exponential Smoothing
- Simple and efficient for trend and seasonality.
- Performed better with lower forecasting errors.

## Conclusion
- Exponential Smoothing slightly outperformed ARIMA for this exchange rate dataset.
- Both models provided reliable forecasts for future exchange rates.
- Time series analysis is highly useful for financial forecasting and trend analysis.
